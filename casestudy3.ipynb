```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class Stack:
    def __init__(self):
        self.top = None

    def push(self, x):
        n = Node(x)
        n.next = self.top
        self.top = n
        print(f"Element {x} successfully pushed into the stack.")

    def pop(self):
        if self.top:
            print(f"Element {self.top.data} is popped from the stack.")
            self.top = self.top.next
        else:
            print("Stack is empty.")


class Queue:
    def __init__(self):
        self.front = self.rear = None

    def enqueue(self, x):
        n = Node(x)
        if not self.rear:
            self.front = self.rear = n
        else:
            self.rear.next = n
            self.rear = n
        print(f"Element {x} successfully added to the queue.")

    def dequeue(self):
        if self.front:
            print(f"Element {self.front.data} is dequeued from the queue.")
            self.front = self.front.next
            if not self.front:
                self.rear = None
        else:
            print("Queue is empty.")

    def display(self):
        p = self.front
        print("Current Queue:", end=" ")
        while p:
            print(p.data, end=" ")
            p = p.next
        print()


s, q = Stack(), Queue()

while True:
    print("\n========== MAIN MENU ==========")
    print("1. Stack Operations")
    print("2. Queue Operations")
    print("3. Exit")
    choice = int(input("Enter your choice: "))

    if choice == 1:
        print("\n------ STACK OPERATIONS ------")
        print("1. Push\n2. Pop\n3. Display")
        op = int(input("Enter operation: "))

        if op == 1:
            n = int(input("\nHow many elements do you want to push? "))
            for i in range(n):
                s.push(int(input(f"Enter element {i+1}: ")))
        elif op == 2:
            s.pop()

    elif choice == 2:
        print("\n------ QUEUE OPERATIONS ------")
        print("1. Enqueue\n2. Dequeue\n3. Display")
        op = int(input("Enter operation: "))

        if op == 1:
            n = int(input("\nHow many elements do you want to enqueue? "))
            for i in range(n):
                q.enqueue(int(input(f"Enter element {i+1}: ")))
            q.display()

        elif op == 2:
            q.dequeue()

        elif op == 3:
            q.display()

    elif choice == 3:
        print("\nThank you! Program ended successfully.")
        break

    else:
        print("Invalid choice!")
```
